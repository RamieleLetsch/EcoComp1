# Processamento de arquivos


```python
# Leitura de arquivo

arquivo001 = open("arquivo001.txt",'r')
```

```python
type(arquivo001)
```

```python
arquivo001.close()
```

```python
arquivo002 = open("arquivo002.txt",'w')

for linha in range(1, 101):
    arquivo002.write('%d\n' % linha)
arquivo002.close()
```

```python
arquivo003 = open('arquivo003.txt','w')

arquivo003.write('casa')

arquivo003.close()
```

## Um outro jeito de usar a função open - gerenciador de contexto with 


```python
with open('arquivo4.txt', 'w') as arquivo4obj:
    arquivo4obj.write('casa')
```

## Criando um csv

```python
with open('arquivo5.csv','w') as arquivo5:
    i = 0
    while i < 11:
        arquivo5.write(f'{i}, {2*i}\n')
        i+=1
```
